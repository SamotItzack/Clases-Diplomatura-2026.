##Actividad 3 - Tipos de datos

### ¿Qué es `type()`?

Cada dato que guardamos en una variable tiene un **tipo**. la función `type()` nos dice qué tipo de datos le pusimos a la variable.

```python
type(27)        # int
type(1.65)      # float
type("Yamila")  # str
type(True)      # bool
```

## Los 4 tipos de datos principales

| Tipo | Significado | Ejemplos |
|---|---|---|
|`int`| Número entero (sin decimales) | `27`, `-5`, `0`|
|`float`| Número decimal (se escribe con punto) | `1.65`, `3.14`|
|`str`| Texto o cadena de caracteres (siempre entre comillas) | `Yamila`|
|`bool`| Booleano: solo puede ser `True` (verdadero) o `False` (falso) | `True`, `False`|

**Importante:** lo que está entre comillas es siempre un `str`, aunque adentro tenga un numero. `27` es un `int`, pero `"27"` es un `str`.

## Operaciones entre tipos

**Números (`int` y `float`):**

- `altura / 2` → la división `/` **siemrpe devuelve un `float`**, aunque los números sean enteros (`10 / 2` da `5.0`).
- `numero * 2` → se puede multiplicar un `float` por un `int`, da igual que en una calculadora.
- `edad2 * altura` → `27 * 1.65 = 44.55`.

**Textos (`str`):**
- `"casa" * 2` → `"casacasa"`: multplicar un texto por un entero lo **repite**.
- `"casa" + "2"` → `"casa 2"`: sumar textos los **une**. Esto se llama **concatenación**. El `2` es un string porque está entre comillas.
- `nombre * 2` → `"YamilaYamila"`.


**Operaciones que dan error:**
- `"casa" + 2` → no se puede sumar un texto con un número.
- `altura * nombre` → no se puede multiplicar un texto por un `float` (no se puede repetir un texto 1.75 veces).


## La función `input()` siempre devuelve `str`

Todo lo que ingresa el usuario mediante `input()` es un **string**, aunque escriba número.

```python
edad = input("¿Cuál es tu edad?")  # el usuario escribe 18
type(edad)                         # str
```
Para poder hacer cuentas hay que **convertir el tipo** (Conversión de tipos):

```python
altura = input("¿Cuál es tu altura?") # escribe 1.75
print(type(altura))                   # <class 'str'>
altura = float(altura)                # conversión a float
type(altura)                          # float
```

Funciones de conversión:
- `int("18")` → `18`
- `float("1.75")` → `1.75`
- `str(18)` → `"18"`

Se puede hacer todo en una linea: `altura = float(input("¿Cuál es tu altura?"))`.

## Boleanos y comparaciones

Los operadores de comparación devuelven un valor tipo `bool`:

| Operador | Significado | Ejemplo | Resultado |
|---|---|---|---|
| `<` | menor que | `50 < 18` | `False` |
| `>` | mayor que | `50 > 18` | `True` |
| `==` | igual a | `50 == 5` | `False` |
| `!=` | distinto de | `50 != 5` | `True` |

**Diferencia entre `=` y `==`:**
- `=` **guarda** un valor en una variable: `edad = 18`
- `==` **compara** dos valores y responde `True` o `False`: `edad == 18`

Ejemplo: `afirmacion = altura > 18` guarda el resultado de la comparación, así que `afirmacion` es de tipo `bool`.

**`True` y `False` también son números:** para Python, `True` vale **1** y `False` vale **0**, por eso se pueden sumar.

```python
True + True + True + True   # 4
True + False                # 1
False + False               # 0
```

Con `int()` se puede convertir un booleano a número: `int(True)` da `1`.

## Errores y cosas a tener en cuenta

- **Una celda solo muestra el resultado de su última línea.** Las otras líneas se calculan igual, pero no se ven (para verlas hay que usar `print()`).
- **`type = 3` es un error típico:** al hacerlo pisamos la función `type()` y pasa a ser el número 3, así que `type(edad)` deja de funcionar. **No hay que usar nombres de funciones de Python** (`type`, `print`, `input`, `int`, `float`, `str`) como nombres de variables.
- **`NameError: name 'nombre2' is not defined`:** Python solo conoce las variables de las celdas que ya se ejecutaron. Si escribimos `nombre2 = 'clara'` pero no ejecutamos esa celda, la variable no existe y al usarla da error.
- **Variables que cambian de tipo:** `edad = 18 < 54` guarda `True`, así que `edad` deja de ser un número y pasa a ser un `bool`.

# 🔴 Clase 3 (Virtual): Condicionales y operadores lógicos

## 1. Repaso

  - **Variable:** espacio para guardar información que puede cambiar
  - **Algoritmo:** conjunto ordenado de pasos a realizar una tarea.
  - **Tipos de datos:** numéricos (entero `int` y real `float`), string (texto, siempre entre comillas) y booleanos (`True` / `False`). Más adelante vienen listas, diccionarios y conjuntos.

## 2. Alternativa condicional

Un condicional permite que el programa **tome decisiones:** según si una condición es verdadera o falsa, ejecuta un bloque u otro.
```python
if condición      #un solo caso
    acción

if condición      #dos caminos
    acción1
else:
  acción2

if condición_1     #varios caminos
    acción1
elif condicion_2:
  accion2
else:
  accion3
```

|En palabras | En Python |
|---|---|
|Si | `if`|
|Sino si | `elif`|
|De lo contrario | `else`|

**Reglas de sintaxis**

  - Después de la condición van **dos putos `:`** (en `if`, `elif` y `else`).
  - Las acciones van **indentadas**, en la línea de abajo.
  - El `else` **no lleva condición**: significa "en cualquien otro caso".
  - Se pueden poner todos los `elif` que haga falta.
  - Python revisa de arriba hacia abajo y entra al **primer** bloque cuya condición sea verdadera. Los demas se saltean

### Operaciones de comparación

Todos devuelven un `bool`.

| Símbolo | Nombre | Ejemplo | Verdadero si... |
|---|---|---|---|
|`<` | Menor que | `x < 5` | x es menor que 5 |
|`>` | Mayor que | `x > 5` | x es mayor que 5 |
|`<=` | Menor o igual que | `x <= 5` | x es menor o igual que 5 |
|`>=` | Mayor o igual que | `x >= 5` | x es mayor o igual que 5 |
|`==` | Igual a | `x == 5` | x es igual a 5 |
|`!=` | Distinto a | `x != 5` | x es distinto a 5 |

- `=` **guarda** un valor (`x = 8`). `==` **compara** (`x == 8`).
- Si `x = 8`: `>5` es `True`, `<5` es `False`, `>=5` es `True`, `<=5` es `False`, `==5` es `False`, `!=5` es `True`.
- Si `x = "cinco"`: `==` y `!=` no dan error. Con `>`, `<`, **COMPLETAR COPIANDO Y PEGANDO ESTA PARTE **

## Actividad 3: Operadores lógicos

Sirven para **combinar dos o más condiciones**. El resultado también es un booleano.

| Operador | Descripción | Ejemplo | En criollo |
|---|---|---|---|
|`and` | `True` si **ambas** son verdaderas | `lluvioso` and `temperatura < 10` | "el clima tiene que estar `lluvioso` y `la temperatura menor que 10`"|
|`or` | `True` si **al menos una** es verdadera | `rojo` or `ROJO` |  "la variable puede ser `rojo` en minuscula o `ROJO` en mayuscula" |
|`not` | **Invierte** el valor | `not x > 5` | "x **no** es mayor a 5" |

Con los dos valores de la diapositiva:

| Expresión | x = 4 | x = 7 |
|---|---|---|
| `x > 5 and x < 10` | `False` | `True` |
| `x > 5 or x < 10` | `True` | `True` |
| `not x > 5` | `True` | `False` |

El `or` de ese ejemplo **siempre** da `True`: cualquier número es mayor a 5 o menor a 10. Una condición mal armada puede ser siempre verdadera o nunca verdadera.

**Error típico:** cada lado de un `and` / `or` tiene que ser una comparación **completa**.


```python
luz == "rojo" or luz == "naranja" # ✅ bien
luz == "rojo" or "naranja" # ❌ mal
```

### Módulos y `random`

  - Un **módulo** es un archivo con funciones, variables y clases que se puede **importar** para reutilizar código de otros programadores.
  - El primer módulo es `random`, que tiene funciones para generar cosas "al azar".
  - Se importa con `import random` y se usa como `random.funcion()`.
  - `random.randint(1,5)` devuelve un número **entero al azar entre 1 y 5**, ambos incluidos. Cada vez que lo ejecutás puede dar un resultado distinto.
  - Para simular una moneda, la idea es pensar cuántos resultados posibles hay (dos) y cómo representarlos con números.
  - Documentación oficial: https://docs.python.org/es/3.13/library/random.html


````python
import random

jugador_1 = input("Nombre Jugador 1")
jugador_2 = input("Nombre Jugador 2")

puntos_1 = 0
puntos_2 = 0

#Elección Jugadores;

while otra_ronda := input("¿Jugar de nuevo? (s/n):").strip().lower() == "s":  #podes hacer una variable que se llame "jugar_siguiente_ronda" y usar la condicion del while como variable y texto.

    turno = random.randint(1,2)
    if turno == 1:
      turno_jugador = jugador_1
      otro = jugador_2
    else:
      turno_jugador = jugador_2
      otro = jugador_1

    eleccion = input(turno_jugador + ", Pon su elección→").lower()


    # Prendas / Retos;
    prenda1 = ("Contar un chiste malo (el más malo que se sepa).")
    prenda2 = ("Hablar con acento de otro país durante la próxima ronda, por ejemplo mexicano o español.")
    prenda3 = ("Hacer 5 sentadillas mientras dice su propio nombre.")
    prenda4 = ("Cantar el estribillo de una canción que elija el otro jugador.")

    #Lógica de la moneda; (1=Cara), (2=Cruz);
    moneda = random.randint(1, 2)

    if moneda == 1:
      resultado = "cara"
    else:
      resultado = "cruz"

    print("¡Salió", resultado + "!")

    #Lógica del ganador;
    if eleccion == resultado:
      ganador = turno_jugador
      perdedor = otro
    else:
      ganador = otro
      perdedor = turno_jugador

    print(ganador, "Gana!!!")

    #Prenda random;
    sorteo = random.randint(1, 4)

    if sorteo == 1:
      prenda = prenda1
    elif sorteo == 2:
      prenda = prenda2
    elif sorteo == 3:
      prenda = prenda3
    else:
      prenda = prenda4

    print(perdedor, "Tu reto es→", prenda)

    #Puntaje;

    if ganador == jugador_1:
      puntos_1 = puntos_1 + 1
    else:
      puntos_2 = puntos_2 + 1

    print("----- PUNTAJE -----")
    print(jugador_1, ":", jugador_2)
    print(puntos_1,  ":", puntos_2)
````

# Actividad 3: Recorrer listas (ciclo `for`)

Un ciclo `for` **repite un bloque de código para cada elemento** de una lista.

###En criollo: **"para cada `e` que haya en `edades`, hacé un print de `e`"**.

# Actividad 4: Moneda recargada

Simulamos tirar una moneda con `random.randint(1,2)` → **1 = cara** y **2 = seca**.

## Juego 4: 10 tiradas

Se tira 10 veces. Si salen más caras gana el jugador 1, si salen más secas gana el jugador 2, y si salen iguales los dos hacen la prenda.

Se usan **dos contadores** (caras y secas) y un `for` con `range(10)`.

````python
import random

jugador_1 = input("Nombre Jugador 1: ")
jugador_2 = input("Nombre Jugador 2: ")
prenda_1 = input("Prenda para el jugador 1: ")
prenda_2 = input("Prenda para el jugador 2: ")

caras = 0
secas = 0

for i in range(10):
  moneda = random.randint(1, 2)
  if moneda == 1:
    caras = caras + 1
  else:
    secas = secas + 1

print("Salieron", caras, "caras y", secas, "secas")

if caras > secas:
  print("Gana", jugador_1)
elif secas > caras:
  print("Gana", jugador_2)
else:
  print("Empate! Los dos hacen la prenda:", prenda_1, "y", prenda_2)
````

## Juego 5: tirar hasta que salga cara

Cada jugador tira hasta que sale cara, y cuenta cuántas secas salieron antes. Esa es la cantidad de saltos en un pie.

### Ciclo `while`

`while` significa **"mientras"**. Repite el bloque **mientras la condición sea `True`** y se frena cuando pasa a ser `False`.

```python
# Estructura
while condición:
  acción
```
**¿Por qué acá es mejor `while` que `for`?** Porque **no sabemos cuántas veces hay que tirar**: puede salir cara a la primera o a la décima. El `for` sirve cuando sabemos la cantidad de vueltas, el `while` cuando depende de una condición.

**Cuidado:** si la condición nunca se vuelve `False`, el ciclo es **infinito** y se queda trabado.

````python
secas = 0
moneda = random.randint(1, 2)    #primera tirada

while moneda == 2:               #mientras salga seca, seguimos tirando
  secas = secas + 1
  moneda = random.randint(1, 2)

print("Salió cara después de", secas, "secas. ¡A hacer", secas, "saltos en un pie!")
````

# Actividad 5: El súper

Compramos chocolates (cada uno vale $2933) **mientras la plata nos alcance** y vemos el vuelto.

La idea: mientras `fondos > precio`, le restamos el precio a los fondos.

````python
precio = 2933
fondos = 15000

while fondos > precio:
  fondos = fondos - precio #resta 5 veces 2933 a los 15000 del fondo.

print('El vuelto será ', fondos, ' pesos')
````

### 1. ¿Cómo saber cuántos chocolates compré?

Agregamos un **contador** que suma 1 en cada vuelta del `while`.

Detalle: usamos `>=` en vez de `>`, porque si me queda justo la plata del chocolate, **sí** lo puedo comprar.

````python
precio = 2933
fondos = 15000
cantidad = 0

while fondos >= precio:
  fondos = fondos - precio
  cantidad = cantidad + 1

print("Compré", cantidad, "chocolates y el vuelto es", fondos, "pesos")
````

### 2. Comprar productos de una lista

Vamos comprando en el orden de la lista, mientras **queden productos** y **alcance la plata**. La variable `orden` es el índice del producto que estoy por comprar y a la vez cuenta cuántos compré.

Dos condiciones con `and`:

* `orden < len(lista_precios)` → todavía quedan productos (si no, daría error al pasarnos de la lista)
* `fondos >= lista_precios[orden]` → me alcanza para el próximo

````python
lista_precios = [900, 7080, 2060, 2933, 2200, 950, 1145, 3000]
lista_productos = ['sal', 'café', 'leche', 'arroz', 'pan', 'yogurt', 'soda', 'atún']
fondos = int(input('Cuanta plata tenés?: '))

orden = 0

while orden < len(lista_precios) and fondos >= lista_precios[orden]:
  fondos = fondos - lista_precios[orden]
  print("Compré", lista_productos[orden])
  orden = orden + 1

print('El vuelto sera ', fondos, 'pesos, alcanzó para ', orden, 'productos')
````

### 4. Extra: estrategia para comprar la mayor cantidad de productos

Comprar **primero los más baratos**. Para eso ordenamos los precios de menor a mayor con `sorted()` y hacemos lo mismo de arriba.

````python
precios_ordenados = sorted(lista_precios)
fondos = int(input('Cuanta plata tenés?: '))
orden = 0

while orden < len(precios_ordenados) and fondos >= precios_ordenados[orden]:
  fondos = fondos - precios_ordenados[orden]
  orden = orden + 1

print('Con esta estrategia alcanzó para', orden, 'productos y sobran', fondos, 'pesos')
````
